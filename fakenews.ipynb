```python
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer

# Assuming 'fake' is a DataFrame with 'text' and 'label' columns
# Splitting the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(fake['text'], fake['label'], test_size=0.2, random_state=42)

# Creating the Bag-of-Words model
vectorizer = CountVectorizer()
X_train_counts = vectorizer.fit_transform(X_train)
X_test_counts = vectorizer.transform(X_test)

# Outputting the shapes of the count vectors
X_train_counts.shape, X_test_counts.shape
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Creating the TF-IDF model
tfidf_vectorizer = TfidfVectorizer()
X_train_tfidf = tfidf_vectorizer.fit_transform(X_train)
X_test_tfidf = tfidf_vectorizer.transform(X_test)

X_train_tfidf.shape, X_test_tfidf.shape
```

```python
from sklearn.neural_network import MLPClassifier


mlp_model = MLPClassifier(hidden_layer_sizes=(100, 100), max_iter=500, random_state=42)
mlp_model.fit(X_train_tfidf, y_train)


y_pred = mlp_model.predict(X_test_tfidf)


accuracy = accuracy_score(y_test, y_pred)
accuracy
```
