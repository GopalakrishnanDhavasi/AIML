```python
import numpy as np

def spherical_cosine_distance(lat1, lon1, lat2, lon2):
    R = 6371.0  # Earth radius in km
    lat1, lon1, lat2, lon2 = map(np.radians, [lat1, lon1, lat2, lon2])
    
    return R * np.arccos(
        np.sin(lat1) * np.sin(lat2) +
        np.cos(lat1) * np.cos(lat2) * np.cos(lon2 - lon1)
    )
```

```python
df['distance'] = df.apply(
    lambda row: spherical_cosine_distance(
        row['Restaurant_latitude'],
        row['Restaurant_longitude'],
        row['Delivery_location_latitude'],
        row['Delivery_location_longitude']
    ), axis=1
)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import MinMaxScaler, OneHotEncoder
import joblib

# Define columns
numeric_features = ['Delivery_person_Age', 'Delivery_person_Ratings', 'distance']
categorical_features = ['Type_of_order', 'Type_of_vehicle']

# Create column transformer
preprocessor = ColumnTransformer(
    transformers=[
        ('num', MinMaxScaler(), numeric_features),
        ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_features)
    ]
)

# Combine preprocessing and model
pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor),
    # ('model', model)  # If using sklearn model. Skip for Keras
])

# Fit only the preprocessor on X
pipeline.fit(X)

# Save the whole pipeline (only 1 file)
joblib.dump(pipeline, 'preprocessing_pipeline.pkl')
```
